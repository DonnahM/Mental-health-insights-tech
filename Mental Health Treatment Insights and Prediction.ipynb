### Data Cleaning and  Preprocessing 

```python
# Standardized column mapping for 2014–2023 OSMI datasets
column_mapping = {
    "age": ["Age", "What is your age?"],
    "gender": ["Gender", "What is your gender?"],
    "country": ["Country", "What country do you *live* in?"],
    "company_size": [
        "no_employees",
        "How many employees does your company or organization have?"
    ],
    "family_history": [
        "family_history",
        "Do you have a family history of mental illness?"
    ],
    "benefits": [
        "benefits",
        "Does your employer provide mental health benefits as part of healthcare coverage?"
    ],
    "care_options": [
        "care_options",
        "Do you know the options for mental health care available under your employer-provided health coverage?"
    ],
    "work_interfere": [
        "work_interfere",
        "If you have a mental health disorder, how often do you feel that it interferes with your work *when* _*NOT*_* being treated effectively (i.e., when you are experiencing symptoms)?*"
    ],
    "anonymity": [
        "anonymity",
        "Is your anonymity protected if you choose to take advantage of mental health or substance abuse treatment resources provided by your employer?"
    ],
    "leave": [
        "leave",
        "If a mental health issue prompted you to request a medical leave from work, how easy or difficult would it be to ask for that leave?"
    ],
    "treatment": [
        "treatment",
        "Have you ever sought treatment for a mental health disorder from a mental health professional?"
    ]
}
  

```

```python
# Function to rename columns based on the mapping
def clean_osmi(df, year, mapping):
    # Start a new clean dataframe
    new_df = pd.DataFrame()
    new_df["year"] = year

    for clean_name, possible_names in mapping.items():
        found = False
        
        for col in possible_names:
            if col in df.columns:
                new_df[clean_name] = df[col]
                found = True
                break
        
        # if no matching column was found in this year's dataset
        if not found:
            new_df[clean_name] = None
    
    return new_df


```

```python
# Clean each year's dataset
df_2014_clean = clean_osmi(df_2014, 2014, column_mapping)
df_2016_clean = clean_osmi(df_2016, 2016, column_mapping)
df_2017_clean = clean_osmi(df_2017, 2017, column_mapping)
df_2018_clean = clean_osmi(df_2018, 2018, column_mapping)
df_2019_clean = clean_osmi(df_2019, 2019, column_mapping)
df_2020_clean = clean_osmi(df_2020, 2020, column_mapping)
df_2021_clean = clean_osmi(df_2021, 2021, column_mapping)
df_2022_clean = clean_osmi(df_2022, 2022, column_mapping)
df_2023_clean = clean_osmi(df_2023, 2023, column_mapping)
```

```python
# Combine all cleaned dataframes into a single dataframe
combined_df = pd.concat([
    df_2014_clean, df_2016_clean, df_2017_clean, df_2018_clean,
    df_2019_clean, df_2020_clean, df_2021_clean, df_2022_clean, df_2023_clean
], ignore_index=True)

```

#### **Gender**

- The *Gender* column had dozens of messy text variations. To make the model readable, all responses were standardized into three groups: Male, Female, and Other.

```python
# Clean Gender Column

# Convert gender values to lowercase and strip spaces
combined_df['gender'] = combined_df['gender'].astype(str).str.lower().str.strip()

# Replace common male variations with "Male"
male_terms = ['male', 'm', 'man', 'cis male', 'male-ish', 'msle', 'mal', 'maile']
combined_df['gender'] = combined_df['gender'].replace(male_terms, 'Male')

# Replace common female variations with "Female"
female_terms = ['female', 'f', 'woman', 'cis female', 'femail']
combined_df['gender'] = combined_df['gender'].replace(female_terms, 'Female')

# Everything else → Other
combined_df.loc[
    ~combined_df['gender'].isin(['Male', 'Female']),
    'gender'
] = 'Other'

# Check distribution after cleaning
combined_df['gender'].value_counts()
```

#### **Country**

- Countries appeared in many different text formats and inconsistent entries. To simplify modeling, country names were standardized, and rare countries were grouped together into an “Other” category to reduce noise.

```python
# --- Clean Country Column ---

# Convert to string, strip spaces, and title-case
combined_df['country'] = combined_df['country'].astype(str).str.strip().str.title()

# Replace common variations
combined_df['country'] = combined_df['country'].replace({
    'Usa': 'United States',
    'United States Of America': 'United States',
    'Uk': 'United Kingdom',
    'U.S.A.': 'United States'
})

# Replace meaningless or unknown entries
combined_df.loc[
    combined_df['country'].isin(['Na', 'None', 'Nan', '', 'Not Applicable']),
    'country'
] = np.nan

# Check unique values after cleaning
combined_df['country'].nunique()
```

- Countries appeared in 81 variations and low-frequency categories. To keep the model stable, the top 10 countries were kept, and the rest were grouped as ‘Other’

```python
# Group countries into top N and label rest 'Other'

# Count frequencies
top_n = 10
country_counts = combined_df['country'].value_counts(dropna=True)

# Create mapping: top N keep as-is, others -> 'Other'
top_countries = country_counts.nlargest(top_n).index.tolist()
combined_df['country_grouped'] = combined_df['country'].where(combined_df['country'].isin(top_countries), other='Other')

# Show results
print("Top countries kept:", top_countries)
print("Unique in country_grouped:", combined_df['country_grouped'].nunique())
print(combined_df['country_grouped'].value_counts().head(15))
```

#### **Company Size**

- Company size had inconsistent formats like ‘6 - 25’, ‘more than 1000’, and even text responses. These were normalized into clean size ranges and converted into ordered categories.

```python
# Normalize company size text and convert to ordered numeric categories

# First, normalize text
combined_df['company_size_raw'] = combined_df['company_size'].astype(str).str.strip()

# Map common variants to canonical labels
size_map = {
    '1-5': '1-5', '6-25': '6-25', '26-100': '26-100', '100-500': '100-500',
    '500-1000': '500-1000', 'more than 1000': '1000+', 'more than 1000 ': '1000+',
    'more than 1000': '1000+', '1000+': '1000+', '6 - 25': '6-25'
}

combined_df['company_size_clean'] = combined_df['company_size_raw'].replace(size_map)

# Define function to infer size from messy text 
def infer_size(x):
    if pd.isna(x) or x.lower() in ['nan', 'none', 'nan']:
        return np.nan
    s = str(x)
    # simple rules
    if re.search(r'1000|more than 1000|>1000', s, flags=re.I):
        return '1000+'
    if re.search(r'500', s):
        return '500-1000'
    if re.search(r'100-500', s):
        return '100-500'
    if re.search(r'26-100', s):
        return '26-100'
    if re.search(r'6-25|6 - 25', s):
        return '6-25'
    if re.search(r'1-5|1 - 5|1 to 5', s):
        return '1-5'
    return np.nan

combined_df['company_size_clean'] = combined_df['company_size_clean'].fillna(combined_df['company_size_raw'].apply(infer_size))

# Create an ordered numeric version for modeling (small -> large)
order = {'1-5':1, '6-25':2, '26-100':3, '100-500':4, '500-1000':5, '1000+':6}
combined_df['company_size_ord'] = combined_df['company_size_clean'].map(order)

# Check results
print(combined_df['company_size_clean'].value_counts(dropna=False))
print("Missing company_size:", combined_df['company_size_ord'].isna().sum())

```

```python
# Add missing mapping to fix missing company size 
combined_df['company_size_clean'] = combined_df['company_size_clean'].replace({
    'More than 1000': '1000+',
    'more than 1000': '1000+'
})

# Re-map
combined_df['company_size_ord'] = combined_df['company_size_clean'].map(order)

# Check results 
print(combined_df['company_size_clean'].value_counts(dropna=False))
print("Missing company_size:", combined_df['company_size_ord'].isna().sum())
```

- Converted to an ordered scale and filled missing values with the median category.

```python
# Fill missing company size ordinals with median to clean company_size_ord
combined_df['company_size_ord'].fillna(combined_df['company_size_ord'].median(), inplace=True)
combined_df['company_size_ord'].isna().sum()
```

#### **Family History**

- This column contained different text formats for yes/no responses. All variations were mapped into a simple numeric binary (1 = Yes, 0 = No) so the model can use it.

```python
# --- Family history to binary (Yes -> 1, No -> 0) ---

# Convert family_history to numeric binary 1/0
combined_df['family_history_clean'] = combined_df['family_history'].astype(str).str.lower().str.strip()
combined_df['family_history_clean'] = combined_df['family_history_clean'].replace({
    'yes':'1', 'no':'0', 'y':'1', 'n':'0', 'true':'1', 'false':'0'
})
combined_df['family_history'] = combined_df['family_history_clean'].map({'1':1, '0':0}).astype('float')
print("family_history value counts:", combined_df['family_history'].value_counts(dropna=False))

```

- To deal with the 736 NaNs, fill missing family history with the mode 

```python
# Impute missing family_history with mode
combined_df['family_history'].fillna(combined_df['family_history'].mode()[0], inplace=True)

# Confirm the imputation 
combined_df['family_history'].value_counts()
```

#### **Benefits. Care Options . Anonymity**

- These three columns recorded Yes/No-style responses, but with many messy variations (e.g., “Y”, “Nope”, “Don’t know”). They were cleaned using the same process:

- Standardized all variations of yes and no into numeric 1/0.

- Marked uncertain answers (like “don’t know”) as missing.

- Filled missing values using the most common response (mode).

```python
# --- Standardize yes/no/dont know type columns ---

# Map Yes->1, No->0, uncertain answers -> NaN for key binary columns and target

binary_cols = ['benefits', 'care_options', 'anonymity']  

def map_yesno(x):
    if pd.isna(x):
        return np.nan
    s = str(x).strip().lower()
    if s in ['yes', 'y', 'true']:
        return 1
    if s in ['no', 'n', 'false']:
        return 0
    if s in ["don't know", "dont know", "not sure", "i don't know", "do not know"]:
        return np.nan  # treat as unknown
    return np.nan

for c in binary_cols:
    combined_df[c + '_bin'] = combined_df[c].apply(map_yesno)


print("benefits_bin counts:\n", combined_df['benefits_bin'].value_counts(dropna=False))
print("care_options_bin counts:\n", combined_df['care_options_bin'].value_counts(dropna=False))
print("anonymity_bin counts:\n", combined_df['anonymity_bin'].value_counts(dropna=False))
```

```python
# --- Impute missing binary columns with mode ---
binary_clean_cols = ['benefits_bin', 'care_options_bin', 'anonymity_bin']

for col in binary_clean_cols:
    combined_df[col].fillna(combined_df[col].mode()[0], inplace=True)

# Check results
for col in binary_clean_cols:
    print(f"{col} after imputation:\n", combined_df[col].value_counts(dropna=False), "\n")
```

### **Treatment**

- Only “Yes” and “No” were kept and converted into 1/0. Where 1 = sought treatment and 0 = did not.

```python
# Treatment (target) mapping: Yes -> 1, No -> 0, others -> NaN
combined_df['treatment_clean'] = combined_df['treatment'].astype(str).str.strip().str.lower().map({
    'yes':1, 'y':1, 'true':1, 'no':0, 'n':0, 'false':0
})

print("treatment_clean counts:\n", combined_df['treatment_clean'].value_counts(dropna=False))
```

- Rows with missing treatment were removed to ensure correct labels.

```python
# Drop rows with missing treatment (target) 
print("Missing treatment before drop:", combined_df['treatment_clean'].isna().sum())

combined_df = combined_df.dropna(subset=['treatment_clean'])

print("New shape after dropping missing target:", combined_df.shape)
print("Treatment distribution after cleaning:\n", combined_df['treatment_clean'].value_counts())
```

#### **Work Interfere**

- Converted “Never/Rarely/Sometimes/Often” into an ordered numeric scale then the missing values were filled using the median, preserving the ordinal nature

```python
# Clean work_interfere into ordinal

# Map work_interfere responses to ordinal values
map_work = {
    'never': 0,
    'rarely': 1,
    'sometimes': 2,
    'often': 3
}

combined_df['work_interfere_clean'] = combined_df['work_interfere'].astype(str).str.lower().str.strip().map(map_work)

# Impute missing with median
combined_df['work_interfere_ord'] = combined_df['work_interfere_clean'].fillna(combined_df['work_interfere_clean'].median())

print("work_interfere_ord:\n", combined_df['work_interfere_ord'].value_counts())
```

#### **Leave** 

- Mapped leave-difficulty categories into an ordered scale from easy → difficult. To keep this order, categories were converted into ordered integers. Missing values were filled with the most common category(mode).

```python
#Clean leave into ordinal

# Map leave responses to ordinal values
map_leave = {
    'very easy': 0,
    'somewhat easy': 1,
    "don't know": 2,
    'somewhat difficult': 3,
    'very difficult': 4
}

combined_df['leave_clean'] = combined_df['leave'].astype(str).str.lower().str.strip().map(map_leave)

# Impute missing leave values with mode
combined_df['leave_ord'] = combined_df['leave_clean'].fillna(combined_df['leave_clean'].mode()[0])

print("leave_ord:\n", combined_df['leave_ord'].value_counts())
```
